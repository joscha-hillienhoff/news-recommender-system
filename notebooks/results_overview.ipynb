# Results overview

Reads the saved validation metrics in `reports/` and compares the recommenders. No dataset is needed: the
numbers come from `reports/results.csv` and `reports/metrics/scores_val_*.txt`, which are written by
`python -m news_recommender_system evaluate <name>`.

```python
from IPython.display import Image
import pandas as pd

from news_recommender_system.config import FIGURES_DIR, METRICS_DIR, REPORTS_DIR
```

## Main comparison

```python
results = pd.read_csv(REPORTS_DIR / "results.csv")
results.assign(co2_mg=results["co2_kg"] * 1e6).drop(columns="co2_kg")
```

## All saved metric files

```python
def read_scores(path):
    return dict(line.split(":") for line in path.read_text().splitlines())


scores = pd.DataFrame(
    {
        path.stem.removeprefix("scores_val_"): read_scores(path)
        for path in sorted(METRICS_DIR.glob("scores_val_*.txt"))
    }
).T.astype(float)
scores
```

The CTR baseline barely reacts to the window length (12 h to one week). The hybrid rows are classifiers that
choose, per impression, which base model ranks it; they are fit and scored on the same impressions and are an
optimistic upper bound.
